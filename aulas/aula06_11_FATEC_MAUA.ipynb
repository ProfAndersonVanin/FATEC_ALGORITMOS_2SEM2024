# Faça um programa que some todos os números digitados pelo usuário até que ele digite 0

```python
soma = 0
num = int(input("Digite um num: "))
while (num != 0):
  soma = soma + num
  num = int(input("Digite um num: "))

print("A soma é: ",soma)
```
